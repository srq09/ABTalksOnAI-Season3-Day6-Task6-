```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
import re
```

```python
text = """
Natural Language Processing (NLP) is a field of Artificial Intelligence.
It helps computers understand human language.
Language processing includes text analysis, sentiment analysis, and more.
"""
```

```python
# Convert to lowercase
text = text.lower()

# Remove special characters
text = re.sub(r'[^a-z\s]', '', text)

# Tokenization (split into words)
words = text.split()
```

```python
word_freq = Counter(words)
```

```python
# Convert to arrays
words_array = np.array(list(word_freq.keys()))
freq_array = np.array(list(word_freq.values()))
```

```python
# Sort frequencies in descending order
sorted_indices = np.argsort(freq_array)[::-1]

# Top 20
top_n = 20
top_words = words_array[sorted_indices][:top_n]
top_freqs = freq_array[sorted_indices][:top_n]
```

```python
plt.figure(figsize=(12,6))  # graph size bada karo

plt.bar(top_words, top_freqs)

plt.xticks(rotation=60, ha='right')  # rotate + align right
plt.title("Top 20 Word Frequencies")
plt.xlabel("Words")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8))

plt.barh(top_words, top_freqs)

plt.title("Top 20 Word Frequencies")
plt.xlabel("Frequency")
plt.ylabel("Words")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# ABTalks (Season-3) - Task Day-06
# Focus: Statistics for NLP
# Problem: Word Frequency Distribution + Top 20 Visualization
# Tool: NumPy
# ============================================

# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
import re

# --------------------------------------------
# STEP 1: Load Text Data (You can replace this with file input)
# --------------------------------------------
text = """
Natural Language Processing (NLP) is a field of Artificial Intelligence.
It helps computers understand human language.
Language processing includes text analysis, sentiment analysis, and more.
NLP is widely used in chatbots, search engines, and recommendation systems.
"""

# --------------------------------------------
# STEP 2: Text Preprocessing
# --------------------------------------------

# Convert to lowercase
text = text.lower()

# Remove special characters and numbers
text = re.sub(r'[^a-z\s]', '', text)

# Tokenization (split into words)
words = text.split()

# Remove Stopwords (important for better NLP analysis)
stopwords = {
    'is','a','the','and','it','of','to','in','on','for','with',
    'this','that','an','as','are','be','by','from','or','at'
}

words = [word for word in words if word not in stopwords]

# --------------------------------------------
# STEP 3: Compute Word Frequency
# --------------------------------------------
word_freq = Counter(words)

# --------------------------------------------
# STEP 4: Convert to NumPy Arrays
# --------------------------------------------
words_array = np.array(list(word_freq.keys()))
freq_array = np.array(list(word_freq.values()))

# --------------------------------------------
# STEP 5: Sort and Extract Top 20 Words
# --------------------------------------------
sorted_indices = np.argsort(freq_array)[::-1]

top_n = 20
top_words = words_array[sorted_indices][:top_n]
top_freqs = freq_array[sorted_indices][:top_n]

# --------------------------------------------
# STEP 6: Visualization (Bar Chart)
# --------------------------------------------
plt.figure(figsize=(10, 5))
plt.bar(top_words, top_freqs)

plt.title("Top 20 Word Frequency Distribution")
plt.xlabel("Words")
plt.ylabel("Frequency")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()

# --------------------------------------------
# STEP 7: Print Output (Optional but good practice)
# --------------------------------------------
print("Top 20 Words with Frequencies:\n")
for word, freq in zip(top_words, top_freqs):
    print(f"{word} : {freq}")
```
